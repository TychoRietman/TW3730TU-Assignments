**(b)** Write down the $O(h^2)$ finite-difference (FD) approximation of the (negative) second derivative operator, including the explicit remainder term

For an internal grid point $x_j$, the second derivative is approximated using the central finite-difference formula. Using Taylor expansions around $x_j$,

$$
u(x_j+h)
=
u(x_j)
+h u'(x_j)
+\frac{h^2}{2}u''(x_j)
+\frac{h^3}{6}u'''(x_j)
+\frac{h^4}{24}u^{(4)}(x_j)
+O(h^5),
$$

and

$$
u(x_j-h)
=
u(x_j)
-h u'(x_j)
+\frac{h^2}{2}u''(x_j)
-\frac{h^3}{6}u'''(x_j)
+\frac{h^4}{24}u^{(4)}(x_j)
-O(h^5).
$$

Adding these equations gives

$$
u(x_j+h)-2u(x_j)+u(x_j-h)
=
h^2u''(x_j)
+\frac{h^4}{12}u^{(4)}(x_j)
+O(h^6).
$$

Therefore,

$$
u''(x_j)
=
\frac{u_{j-1}-2u_j+u_{j+1}}{h^2}
-\frac{h^2}{12}u^{(4)}(x_j)
+O(h^4).
$$

Since the problem contains the negative second derivative,

$$
\boxed{
-u''(x_j)
=
\frac{-u_{j-1}+2u_j-u_{j+1}}{h^2}
+\frac{h^2}{12}u^{(4)}(x_j)
+O(h^4)
}
$$

Hence, the finite-difference approximation is

$$
\boxed{
-u''(x_j)
\approx
\frac{-u_{j-1}+2u_j-u_{j+1}}{h^2}
}
$$

with truncation error of order $O(h^2)$.

**(b)** Use `np.diag()` to construct \(A\) from its three diagonals for any given \(n\).

```python
main_diag = 2 * np.ones(n - 1)
off_diag = -1 * np.ones(n - 2)

A = (
    np.diag(main_diag)
    + np.diag(off_diag, 1)
    + np.diag(off_diag, -1)
)
```

### 7(a)

The global error is computed using the discrete $L_2$ error defined as

$$
E =
\sqrt{
\frac{1}{n-1}
\sum_{i=1}^{n-1}
\left(u_i-u_i^{ex}\right)^2
}.
$$

Since the boundary values are imposed exactly, only the interior grid points are used in the error computation.
For $n=5$, the global errors are approximately

$$
E_1 \approx 2.22\times10^{-16},
$$

and

$$
E_2 \approx 5.51\times10^{-2}.
$$

The error for $u_1$ is essentially zero and is only caused by floating-point round-off. This occurs because $u_1^{ex}$ is a cubic polynomial. The centered second-order finite-difference approximation of the second derivative is exact for cubic polynomials since the fourth derivative is zero,

$$
\frac{u_{i-1} +2{u_i} - u_{i+1}}{h^2} = u_i'' + \frac{h^2}{12} u^{(4)}_i + ...
$$

$$
\frac{d^4u_1^{ex}}{dx^4}=0.
$$

For $u_2$, the exact solution is a fourth-order polynomial, so its fourth derivative is nonzero. Therefore, the finite-difference approximation introduces a truncation error, resulting in a nonzero global error,

$$
E_2 \approx 0.0551.
$$

Since the central finite-difference approximation is second-order accurate, the global error is expected to behave as

$$
E=O(h^2).
$$






**(b)** Study the rate of convergence of the Finite-Difference Method by changing \(n\). Display the logarithm of the global error as a function of \(n\). Explain your results.

```python
n_values = np.array([5, 10, 20, 40, 80, 160])

def global_error(n, func, u_exact):
    h = 3.0 / n
    x = np.linspace(0.0, 3.0, n + 1)

    # Finite-difference matrix
    A = (np.diag(2*np.ones(n-1))
         + np.diag(-np.ones(n-2), 1)
         + np.diag(-np.ones(n-2), -1))

    # RHS including boundary conditions u(0) = u(3) = 1
    rhs = h**2 * func(x[1:-1])
    rhs[0] += 1
    rhs[-1] += 1

    # Numerical solution including boundary values
    u = np.concatenate(([1.0], np.linalg.solve(A, rhs), [1.0]))

    # Global L2 error on the interior points
    return np.sqrt(np.mean((u[1:-1] - u_exact(x[1:-1]))**2))


error1_values = np.array([global_error(n, func1, u1_exact) for n in n_values])
error2_values = np.array([global_error(n, func2, u2_exact) for n in n_values])

# Determine convergence order for u2
slope = np.polyfit(np.log(n_values), np.log(error2_values), 1)[0]
p = -slope

print("n values:", n_values)
print("u1 errors:", error1_values)
print("u2 errors:", error2_values)
print(f"Observed convergence order p = {p:.3f}")



plt.figure(figsize=(7, 5))

plt.loglog(n_values, error2_values, 'ro-', label=r'$E_2$')

# Reference line with expected second-order convergence
reference = error2_values[0] * (n_values[0] / n_values)**2
plt.loglog(n_values, reference, 'k--', label=r'$O(n^{-2})$')

plt.xlabel(r'$n$')
plt.ylabel(r'Global error $E$')
plt.title(r'Convergence of the FDM solution')
plt.legend()
plt.grid(True, which='both')

plt.show()
```

### 7(b)

The convergence of the finite-difference method is studied by repeating the computation for increasing values of $n$. Since the interval $[0,3]$ is divided into $n$ equal sub-intervals, the grid spacing is

$$
h=\frac{3}{n}.
$$

The finite-difference approximation used for the second derivative is second-order accurate. Its truncation error is therefore of order

$$
O(h^2).
$$

This means that the global discretization error is expected to behave approximately as

$$
E \approx C h^2,
$$

where $C$ is a constant independent of $h$.

Because

$$
h=\frac{3}{n},
$$

we obtain

$$
E \approx C\left(\frac{3}{n}\right)^2.
$$

Therefore,

$$
E \propto n^{-2}.
$$

More generally, if the error behaves as

$$
E \propto n^{-p},
$$

then taking logarithms gives

$$
\log E = \log C - p\log n.
$$

Hence, a log-log plot of the global error $E$ against $n$ should approximately form a straight line with slope

$$
-p.
$$

For a second-order finite-difference method, the expected value is therefore

$$
p\approx 2,
$$

and the slope of the log-log convergence curve should be approximately $-2$.

This also means that if $n$ is doubled, the grid spacing $h$ is halved. Since the error scales with $h^2$,

$$
E_{\text{new}}
\approx
\left(\frac{1}{2}\right)^2 E_{\text{old}}
=
\frac{1}{4}E_{\text{old}}.
$$

Thus, doubling the number of sub-intervals should reduce the error by approximately a factor of four. The observed convergence order being close to $p=2$ therefore confirms the expected second-order accuracy of the centered finite-difference approximation.**Answer here....**
