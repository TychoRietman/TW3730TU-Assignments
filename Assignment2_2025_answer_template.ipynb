## Numerical solution of the one-dimensional Poisson’s equation with the Finite-Difference Method

**Numerical Analysis For PDE’s (WI3730TU)**

Consider the following boundary-value problem:

$$
-\frac{d^2u_i}{dx^2}=f_i,\qquad x\in(0,3);
\tag{1}
$$

$$
u_i(0)=1,\qquad u_i(3)=1,\qquad i=1,2;
$$

with the source functions

$$
f_1(x)=3x-2,\qquad f_2(x)=x^2+3x-2,\qquad x\in[0,3].
\tag{2}
$$


### 1.

Find the exact solutions $u_1^{\mathrm{ex}}(x)$ and $u_2^{\mathrm{ex}}(x)$ of the problem (1) corresponding to $f_1(x)$ and $f_2(x)$ given in (2).

**Answer here....**

For $i=1$,
$$
\frac{d^2u_1}{dx^2}=-3x+2.
$$

Integrating once gives $\frac{du_1}{dx}=-\frac{3}{2}x^2+2x+C_1.$ Integrating again yields, $u_1(x)=-\frac{1}{2}x^3+x^2+C_1x+C_2.$
Using $u_1(0)=1$ gives $C_2=1$. Using $u_1(3)=1$ gives $-\frac{1}{2}(3)^3+(3)^2+3C_1+1=1,$ and therefore $C_1=\frac{3}{2}.$ Hence,

$$
\boxed{
u_1^{\mathrm{ex}}(x)
=
-\frac{1}{2}x^3+x^2+\frac{3}{2}x+1
}
$$

For $i=2$,
$$
\frac{d^2u_2}{dx^2}=-x^2-3x+2.
$$
Integrating yields $\frac{du_2}{dx}=-\frac{1}{3}x^3-\frac{3}{2}x^2+2x+C_1.$ Integrating again, $u_2(x)=-\frac{1}{12}x^4-\frac{1}{2}x^3+x^2+C_1x+C_2.$ Using $u_2(0)=1$ gives $C_2=1.$ Using $u_2(3)=1$. Therefore, $-\frac{81}{12}-\frac{27}{2}+9+3C_1=0,$ which gives $C_1=\frac{15}{4}.$ Hence,
$$
\boxed{
u_2^{\mathrm{ex}}(x)
=
-\frac{1}{12}x^4-\frac{1}{2}x^3+x^2+\frac{15}{4}x+1
}
$$







### 2.

Discretize the problem (1) using the Finite-Difference Method (FDM) on a uniform grid obtained by dividing the \([0,3]\) interval into \(n=5\) sub-intervals of equal length.

**(a)** What is the step size \(h\)? How many internal and boundary points do you get? How many unknowns does your numerical problem have?

The interval $[0,3]$ is divided into $n=5$ equal sub-intervals. Therefore, the step size is

$$

**(b)** Write down the $O(h^2)$ finite-difference (FD) approximation of the (negative) second derivative operator, including the explicit remainder term

For an internal grid point $x_j$, the second derivative is approximated using the central finite-difference formula. Using Taylor expansions around $x_j$,

$$
u(x_j+h)
=
u(x_j)
+h u'(x_j)
+\frac{h^2}{2}u''(x_j)
+\frac{h^3}{6}u'''(x_j)
+\frac{h^4}{24}u^{(4)}(x_j)
+O(h^5),
$$

and

$$
u(x_j-h)
=
u(x_j)
-h u'(x_j)
+\frac{h^2}{2}u''(x_j)
-\frac{h^3}{6}u'''(x_j)
+\frac{h^4}{24}u^{(4)}(x_j)
-O(h^5).
$$

Adding these equations gives

$$
u(x_j+h)-2u(x_j)+u(x_j-h)
=
h^2u''(x_j)
+\frac{h^4}{12}u^{(4)}(x_j)
+O(h^6).
$$

Therefore,

$$
u''(x_j)
=
\frac{u_{j-1}-2u_j+u_{j+1}}{h^2}
-\frac{h^2}{12}u^{(4)}(x_j)
+O(h^4).
$$

Since the problem contains the negative second derivative,

$$
\boxed{
-u''(x_j)
=
\frac{-u_{j-1}+2u_j-u_{j+1}}{h^2}
+\frac{h^2}{12}u^{(4)}(x_j)
+O(h^4)
}
$$

Hence, the finite-difference approximation is

$$
\boxed{
-u''(x_j)
\approx
\frac{-u_{j-1}+2u_j-u_{j+1}}{h^2}
}
$$

with truncation error of order $O(h^2)$.
