## **Problem Statement**
Consider a finite, simply-connected open domain $\Omega \subset \mathbb{R}^{2}$ with a smooth boundary $\partial\Omega$ and the outward normal vector $\mathbf{n}$. A substance is produced in $\Omega$ at a local rate $f(x,y,t) > 0$. The produced substance diffuses inside $\Omega$ with a diffusion coefficient $k(x,y) > 0$, where $(x,y) \in \overline{\Omega} = \Omega \cup \partial\Omega$.

### **Question 1: Global Conservation Law & Dimensional Analysis (2 p)**
Denoting by $c(x,y,t)$ $[\text{m}^{-2}]$ the concentration of the substance and by $\mathbf{J}(x,y,t)$ its flux density, write down the global conservation law. What are the physical units of $f$ and $\mathbf{J}$?

*Type your answer and unit derivation below:* 

We can express the total amount of substance inside the open domain $\Omega$ with area element $dS$ as

$$
M(t)=\iint_{\Omega} c(x,y,t)dS.
$$

Its amount can change for two reasons. Material can cross the boundary, or the quantity can be created or destroyed internally. If $\Phi$ denotes the outward flux and $R$ the total rate of internal production, conservation gives

$$
\frac{d}{dt}\iint_{\Omega} c(x,y,t)dS=-\Phi+R.
$$

The negative sign in front of $\Phi$ is important because $\Phi$ is defined as outward flux. A positive outward flux reduces the amount contained inside the domain.

Now let $\mathbf{J}$ denote the flux density and $f(x,y,t)>0$ the local source density. Since $\Omega\subset\mathbb{R}^2$, the total production rate is

$$
R(t)=\iint_{\Omega} f(x,y,t)dS.
$$

Let $\mathbf{n}$ denote the outward unit normal to the boundary $\partial\Omega$. The outward normal flux density is

$$
\mathbf{J}\cdot\mathbf{n},
$$

and therefore the total outward flux through the boundary is

$$
\Phi(t)=\int_{\partial\Omega}\mathbf{J}\cdot\mathbf{n}ds.
$$

Substituting these expressions into the conservation equation gives the global conservation law

$$
\frac{d}{dt}\iint_{\Omega}c(x,y,t)dS
=
-\int_{\partial\Omega}\mathbf{J}(x,y,t)\cdot\mathbf{n}ds
+
\iint_{\Omega}f(x,y,t)dS
$$

Since the concentration has units

$$
[c]={m}^{-2},
$$

and $dS$ has units ${m}^2$, the total amount $M(t)$ is dimensionless. Therefore,

$$
\left[\frac{dM}{dt}\right]={s}^{-1}.
$$

For the source term, $[f][dS]={s}^{-1}$ so

$$
[f]={m}^{-2}{s}^{-1}.
$$

For the flux term, $ds$ has units ${m}$ and $\mathbf{n}$ is dimensionless. Hence,

$$
[{J}][ds]={s}^{-1},
$$

which gives
$$
[\mathbf{J}]={m}^{-1} {s}^{-1}.
$$


### **Question 2: Local Conservation Law Derivation (2 p)**
Derive the local conservation law for $(x,y) \in \Omega$ by applying the Divergence Theorem to your global conservation law under the assumption that it holds for any sub-domain of $\Omega$ for all $t > 0$.

*Type your step-by-step derivation below:* 

From the global conservation law derived in Step 1,

$$
\frac{d}{dt}\iint_{\Omega} c(x,y,t)dS
=
-\int_{\partial\Omega}\mathbf{J}\cdot\mathbf{n}ds
+
\iint_{\Omega} f(x,y,t)dS.
$$

Since the domain $\Omega$ is fixed in time, the time derivative can be moved inside the integral,

$$
\iint_{\Omega}\frac{\partial c}{\partial t}dS
=
-\int_{\partial\Omega}\mathbf{J}\cdot\mathbf{n}ds
+
\iint_{\Omega}fdS.
$$

The divergence theorem in two dimensions states that

$$
\int_{\partial\Omega}\mathbf{J}\cdot\mathbf{n}ds
=
\iint_{\Omega}\nabla\cdot\mathbf{J}dS.
$$

Substituting this into the global conservation law gives

$$
\iint_{\Omega}\frac{\partial c}{\partial t}dS
=
-\iint_{\Omega}\nabla\cdot\mathbf{J}dS
+
\iint_{\Omega}fdS.
$$

Bringing all terms into one integral gives

$$
\iint_{\Omega}
\left(
\frac{\partial c}{\partial t}
+
\nabla\cdot\mathbf{J}
-
f
\right)
dS
=
0.
$$

Since conservation must hold for every arbitrary control region contained in $\Omega$, the integrand itself must vanish. Therefore,

$$

\frac{\partial c}{\partial t}
+
\nabla\cdot\mathbf{J}
=
f

$$

which is the local conservation law. In two dimensions,

$$
\mathbf{J}
=
\begin{pmatrix}
J_x\\
J_y
\end{pmatrix},
$$
so

$$
\nabla\cdot\mathbf{J}
=
\frac{\partial J_x}{\partial x}
+
\frac{\partial J_y}{\partial y}.
$$

Hence the local conservation law can also be written as
$$
\frac{\partial c}{\partial t}
+
\frac{\partial J_x}{\partial x}
+
\frac{\partial J_y}{\partial y}
=
f.
$$


### **Question 3: Transport Governing Equation & Diffusion Units (2 p)**
Use Fick's law of diffusion to formulate the partial differential equation describing the local transport of $c$ for $(x,y) \in \Omega$. Deduce the physical units of the diffusion coefficient $k$.

*Type your formulation and unit analysis below:* 


From Question 2, the local conservation law is

$$
\frac{\partial c}{\partial t}+\nabla\cdot\mathbf{J}=f.
$$

Fick's law of diffusion is

$$
\mathbf{J}=-k\nabla c.
$$

The negative sign indicates that the substance diffuses in the direction of decreasing concentration, so from high concentration to low concentration.
Substituting Fick's law into the local conservation law gives

$$
\frac{\partial c}{\partial t}
+
\nabla\cdot\left(-k\nabla c\right)
=
f.
$$

Therefore,

$$

\frac{\partial c}{\partial t}
-
\nabla\cdot\left(k\nabla c\right)
=
f

$$

for $(x,y)\in\Omega$. Equivalently,

$$
\frac{\partial c}{\partial t}
=
\nabla\cdot\left(k\nabla c\right)+f
$$

Assuming that $k$ is constant, $\nabla\cdot\left(k\nabla c\right)$ reduces to

$$
\nabla\cdot(k\nabla c)
=
k\Delta c.
$$

Writing out the Laplacian term, in two dimensions,
$$
\Delta c
=
\frac{\partial^2 c}{\partial x^2}
+
\frac{\partial^2 c}{\partial y^2}.
$$

Hence, the final PDE is

$$

\frac{\partial c}{\partial t}
=
k\left(
\frac{\partial^2 c}{\partial x^2}
+
\frac{\partial^2 c}{\partial y^2}
\right)
+
f

$$

Finally, from Fick's law, $\mathbf{J}=-k\nabla c.$. From Question 1, $[\mathbf{J}]=\mathrm{m}^{-1}\mathrm{s}^{-1},$ and since

$$
[c]={m}^{-2},
$$

we have

$$
[\nabla c]
=
{m}^{-3}.
$$

Therefore,

$$
[k]
=
\frac{[\mathbf{J}]}{[\nabla c]}
=
\frac{{m}^{-1}{s}^{-1}}{{m}^{-3}}
=
{{m}^2 {s}^{-1}}.
$$


### **Question 4: Initial Conditions (1 p)**

### **Question 5: Boundary Conditions (1 p)**
Assuming that the substance flows outwards through the boundary $\partial\Omega$ at a local rate (normal flux density) proportional to the local concentration of the substance at $\partial\Omega$, with a coefficient of proportionality $\alpha(x,y) > 0$, write down the boundary condition on $c$. What are the physical units of $\alpha$?

*Type your boundary condition and unit analysis below:* 

We know from Flick's law that the flux density is $\bold{J} = -k\nabla c$. The outward/normal flux density is therefore,
$$
\bold{J} \cdot \bold{n} = -k \nabla c \cdot \bold{n} = -k D_{n} c
$$
If this is proportional to the local concentration with a coefficient of proportionality $\alpha(x,y)>0$, then

$$
-kD_n c=\alpha c,
$$
We can now write the Robin boundary condition,

$$
kD_n c+\alpha c=0,
\qquad (x,y)\in\partial\Omega
$$

We can derive the physical unit  $\alpha$ from the relation $\bold{J}\cdot \bold{n} = \alpha c $, then
$$
[\alpha] = \frac{[J]}{[c]} = \frac{m^{-1}s^{-1}}{m^{-2}} = ms^{-1}
$$

### **Question 6: Steady-State Analysis & Uniqueness Proof (4 p)**
Formulate the steady-state boundary-value problem assuming $\frac{\partial c}{\partial t} = 0$, and analyze it by answering the following parts:

**(a)** What is the classification/type of the resulting governing PDE? What is the type of the boundary condition? *(1 p)*  
**(b)** Let $k = 1$ and suppose there is a solution $c$ to the steady-state boundary-value problem satisfying $c \in C^{2}(\Omega) \cap C(\overline{\Omega})$. Is it unique? To establish this fact, you also need to prove the corresponding Theorem. *(3 p)*

*Type your steady-state formulation, classification, and full mathematical uniqueness proof below:* 

(a) The steady-state PDE is

$$
-\nabla\cdot(k\nabla c)=f
\qquad \text{in } \Omega.
$$

From Question 5, the boundary condition is a Robin condition, since it contains a Dirichlet and Neumann term.

$$
kD_n c+\alpha c=0
\qquad \text{on } \partial\Omega.
$$

Hence, the steady-state Robin boundary-value problem is

$$

\begin{cases}
-\nabla\cdot(k\nabla c)=f, & (x,y)\in\Omega\\
kD_n c+\alpha c=0, & (x,y)\in\partial\Omega.
\end{cases}
$$

Expanding the differential operator gives

$$
-\nabla\cdot(k\nabla c)
=
-\frac{\partial}{\partial x}
\left(k\frac{\partial c}{\partial x}\right)
-
\frac{\partial}{\partial y}
\left(k\frac{\partial c}{\partial y}\right).
$$

Using the product rule, assuming $k$ is constant

$$
-\nabla\cdot(k\nabla c)
=
-kc_{xx}-kc_{yy} = -k \Delta{c}.
$$

The highest-order part is $-kc_{xx}-kc_{yy},$ so the coefficient matrix is

$$
A=
\begin{pmatrix}
-k & 0\\
0 & -k
\end{pmatrix}.
$$

Its eigenvalues are $\lambda_1=\lambda_2=-k.$. Since $k>0$, both eigenvalues are nonzero and have the same sign. Therefore, the PDE is elliptic and has the form,

$$
-k \Delta c = f
$$

(b)

Since $k=1$ the PDE becomes the Poisson PDE $-\Delta c =f(x,y)$. Assume that $c_1$ and $c_2$ are two solutions and define $w=c_1-c_2.$ Since

$$
\Delta w
=
\Delta(c_1-c_2)
=
\Delta c_1-\Delta c_2
=
(-f)-(-f)
=
0,
$$

we obtain

$$
-\Delta w=0
\qquad \text{in }\Omega.
$$

For the boundary conditions,

$$
D_n c_1+\alpha c_1=0
$$
and
$$
D_n c_2+\alpha c_2=0.
$$

Subtracting gives

$$
D_n(c_1-c_2)+\alpha(c_1-c_2)=0.
$$

Since $w=c_1-c_2$,

$$
D_n w+\alpha w=0.
$$

Hence $w$ satisfies

$$
\begin{cases}
-\Delta w=0, & \text{in }\Omega,\\
D_n w+\alpha w=0, & \text{on }\partial\Omega.
\end{cases}
$$

To prove uniqueness, it is therefore sufficient to show that $w=0.$


Since $-\Delta w=0$, we have $\Delta w=0$, so $w$ is harmonic in $\Omega$. By the maximum and minimum principles, if $w$ is nonconstant, its maximum and minimum on $\overline{\Omega}$ are attained on $\partial\Omega$.

Suppose that $w$ attains a positive maximum

$$
M=\max_{\overline{\Omega}}w>0
$$

at some point $(x_0,y_0)\in\partial\Omega$. At a boundary maximum, the outward normal derivative satisfies

$$
D_n w(x_0,y_0)\geq 0.
$$

However, from the Robin boundary condition,

$$
D_n w=-\alpha w.
$$

Therefore,

$$
D_n w(x_0,y_0)
=
-\alpha(x_0,y_0)M<0,
$$

since $\alpha>0$ and $M>0$. This is a contradiction. Hence,

$$
\max_{\overline{\Omega}}w\leq 0.
$$

Now suppose that $w$ attains a negative minimum

$$
m=\min_{\overline{\Omega}}w<0
$$

at some point $(x_1,y_1)\in\partial\Omega$. At a boundary minimum,

$$
D_n w(x_1,y_1)\leq 0.
$$

But the boundary condition gives

$$
D_n w(x_1,y_1)
=
-\alpha(x_1,y_1)m>0,
$$

since $\alpha>0$ and $m<0$. This is again a contradiction. Therefore,

$$
\min_{\overline{\Omega}}w\geq 0.
$$

Combining the two results,

$$
\max_{\overline{\Omega}}w\leq 0
\qquad\text{and}\qquad
\min_{\overline{\Omega}}w\geq 0,
$$

which implies

$$
w=0
\qquad\text{in }\overline{\Omega}.
$$

If $w$ is constant, then $D_n w=0$. The boundary condition becomes

$$
\alpha w=0.
$$

Since $\alpha>0$, this again implies $w=0$.

Hence,

$$
u_1-u_2=w=0,
$$

so

$$
u_1=u_2.
$$

Therefore, the Robin boundary-value problem has at most one solution.
